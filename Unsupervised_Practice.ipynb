K-Means Clustering

```python
from sklearn.cluster import KMeans

data = [
    [600,7],
    [700,6],
    [500,4],
    [650,8],
    [1000,10],
    [800,9],
    [1200,14],
    [1500,15],
    [1100,12],

]

model = KMeans(n_clusters=3, random_state=42)

model.fit(data)

print(model.labels_)
```

Elbow Method

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = [[600,7],
    [700,6],
    [500,4],
    [650,8],
    [1000,10],
    [800,9],
    [1200,14],
    [1500,15],
    [1100,12],

]

inertia_values = []
k_range = range(1,7)


# calculate inertia for k =1 to 6

for k in k_range:
  model = KMeans(n_clusters=k, random_state=42)
  model.fit(data)
  inertia_values.append(model.inertia_)

# print inertia values

for k, inertia in zip(k_range, inertia_values):
  print(f"k = {k} | Inertia = {inertia:,.2f}")


#plot the elbow curve
plt.figure(figsize=(6,4))
plt.plot(k_range, inertia_values, marker='o')
plt.title("Elbow Curve")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.grid(True)
```

Hierarchical Clustering

```python
from sklearn.cluster import AgglomerativeClustering
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt

data = [
    [90, 80],
    [85, 90],
    [92, 85],
    [60, 62],
    [55, 62],
    [55, 63],
    [35, 45],
    [36, 39]
]

model = AgglomerativeClustering(n_clusters=3)

labels = model.fit_predict(data)

print(labels)


linkage_matrix = sch.linkage(data, method='ward')
sch.dendrogram(linkage_matrix)
plt.title("Dendrogram")
plt.show()
```

PCA - Dimensionality Reduction

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 4 features -> 2 components
data = [[85, 80, 95, 10], [87, 90, 92, 12], [45, 40, 65, 5], [47, 52, 60, 6]]

# 1. Scale
scaled = StandardScaler().fit_transform(data)

# 2. Reduce
reduced = PCA(n_components=2).fit_transform(scaled)

print(reduced)
```

t-SNE (t-Distributed Stochastic Neighbor Embedding)

```python
import numpy as np
from sklearn.manifold import TSNE

data = np.array([
    [1, 2, 3, 4],
    [1, 2, 3, 5],
    [10, 11, 12, 13],
    [10, 11, 12, 14],
    [20, 21, 22, 23],
    [20, 21, 22, 24]
])

# Set perplexity to a value smaller than the number of rows (e.g., 2)
model = TSNE(n_components=2, perplexity=2, random_state=42)

result = model.fit_transform(data)

print(result)
```
