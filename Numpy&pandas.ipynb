```python
import numpy as np

marks = np.array([[10,20,30,],[20,30,40],[15,35,35],[10,25,30],[15,20,30]])
print("Student marks:",marks)


print("average of subject-wise")
 # x axis=0(col)   axis=1 (row)
print(np.mean(marks, axis=1)) # row


```

```python

print("\nHighest mark:", np.max(marks))
print("\nLowest mark:", np.min(marks))
```

```python
average = np.mean(marks, axis=1)
for i in range(len(average)):
  if average[i] > 28:
    print("student",i)
```

```python
## Pandas

import pandas as pd

data = {
    "Employee": ["Subash","Sai","Pavi","Nisha","Maha"],
    "Department" : ["Electronic","IT","CSE","Finance","Design"],
    "Salary" : [90000,80000,70000,60000,50000]
}

df = pd.DataFrame(data)
print(df)
```

```python
print("\nIT Department")
print(df[df["Department"] == "IT"])
```

```python
print("\nHighest Salary")
print(df[df["Salary"] == df["Salary"].max()])
```

```python
print("\nLowest Salary")
print(df[df["Salary"] == df["Salary"].min()])
```

```python
print("\n 10% Increment")

df["Updated Salary"] = df["Salary"]*1.10

print(df)
```

```python
## Matplotlib

import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [1500,1000,2500,2000,1300]

import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months, sales, marker='o')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```
